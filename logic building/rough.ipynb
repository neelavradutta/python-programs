```python
nums = ['4', '8', '9', '100']
count = 0
for i in range(len(nums)):
    num = int(nums[i])  # Convert string to int for operations
    if num % 3 != 0:
        if num > 3:
            nums[i] = str(num - 1)
            count += 1
        elif num < 3:
            nums[i] = str(num + 1)
            count += 1
        elif num == 1:
            nums[i] = '0'
            count += 1
    else:
        pass

print(count)
```

```python
s="fdhf%S5 145^d"
digi=False
lo=False
up=False
spe=False
for i in range(len(s)):
    if s[i].isdigit()==True:
        digi=True 
    elif s[i].islower()==True:
        lo=True 
    
    elif s[i].isupper()==True:
        up=True 
    else:
        spe=True       
        
if digi==True and  lo==True and  up==True and  spe==True:          
    print("Valid")
else:
    print("Not valid")
```
