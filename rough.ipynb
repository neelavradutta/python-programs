```python

n=146
temp=n
sum=0
while n>0:
    a=n%10
    c=1
    for i in range(a,0,-1):
        c=c*i
    sum=sum+c
    n=n//10
if sum==temp:
    print("strong")
else:
    print("not strong")
```

```python
n=[16,17,4,3,5,2]
leader=[]
for i in range(len(n)-1):
    flag=True
    for j in range(i+1,len(n)):
        if n[i]<n[j]:
            flag=False
            break
            
    if flag==True:
        leader.append(n[i])

print(leader)
```
