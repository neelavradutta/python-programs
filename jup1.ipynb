```python
allowed = "ab"
words = ["ad","bd","aaab","baa","badab"]
count=0
for i in range(len(words)):
    if set(words[i]).issubset(set(allowed)):
        count=count+1

print(count)
        
        
```

```python
s = "abccccdd"
max_len = 0
        
for i in range(len(s)):
    for j in range(1,len(s)):
        s1 = s[i:j+1]
                
        if s1 == s1[::-1]:
            max_len = max(max_len, len(s1))
print(max_len)
```

```python
s="AbczcdftgZz"
s1=""
for i in range(len(s)):
    if s[i]!="Z" and s[i]!="z":
        s1=s1+chr(ord(s[i])+1)
    else:
        if s[i]=="Z":
            s1=s1+"A"
        else:
            s1=s1+"a" 
print(s1)
```
